Tuple of NumPy arrays: `(x_train, y_train), (x_test, y_test)`.

  `(60000, 28, 28)`, containing the training data.

**`y_train`**: `uint8` NumPy array of labels (integers in range 0-9)

**`x_test`**: `uint8` NumPy array of grayscale image data with shapes

**`y_test`**: `uint8` NumPy array of labels (integers in range 0-9)
  with shape `(10000,)` for the test data.

### Build the Model using TensorFlow 2.0 :


We will import the necessary layers to build the model. The Neural Network is constructed from 3 type of layers:

- **Input layer** — initial data for the neural network.

- **Hidden layers** — intermediate layer between input and output layer and place
where all the computation is done.

- **Output layer** — produce the result for given inputs

````python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Flatten, Dense
````

**Flatten()** :

- The Flatten Layer is used to convert the multi-dimensinal data into 1D.

- is used as the input layer to convert the data into a 1-dimensional array for input it to the next layer.

- Flatten simply reshapes the data.

- It does not learn anything.

- It does not change the values.

- It only changes the dimensions (shape).

-  Our image 2D image will be converted to a single 1D column.
input_shape = (28,28) because the size of our input image is 28x28.


**Dense()** :

   - A Dense layer is a layer where each input is connected to each neuron in the layer.
   
   - It is called dense because the connections are dense (complete).
   
   - layer is the regular deeply connected neural network layer.

   - It is most common and frequently used layer.
    
   - We have a dense layer with 128 neurons with activation function relu.
    
   - The rectified linear activation function or ReLU for short is a piecewise linear function that will output the input directly if it is positive, otherwise, it will output zero.

- The **output layer** is a dense layer with 10 neurons because we have 10 classes.

- The activation function used is **softmax**.

- Softmax converts a real vector to a vector of categorical probabilities.

- The elements of the output vector are in range (0, 1) and sum to 1.

- Softmax is often used as the activation for the last layer of a classification network because the result could be interpreted as a probability distribution.

### Step-4 : Modelling(Model Development) : 

![Neural Network Model Development in which different Layer's are there](Neural_Network_Dev_img.jpg)

````python
model = Sequential()

# Our dataset is of images so it is 2D dataset, but the model can't proces or can't trained on 2D data, so need to convert into 1D.
# So the Flatten Layer is used to convert the multi-dimensinal data into 1D.

# Input layer starts using Flatten(), size of input image=(28,28) which will convert in 1D using Flatten().
model.add(Flatten(input_shape=(28,28)))   # 28*28 means 784 pixel values as input neurons for gray image. If RGB image then neaurons will be 784 multiple by 3 means 2352 input neurons.

# How many hidden layers do you want => we want one hidden layer with 128 neurons in it.
# We have to use Relu as the activation function in Hidden layers because it will avoid the vanish gradients problem
    # (means if we use other activation function, it will gives output in between 0 to 1. Means gradient will be less,
    #  while going through each hidden layer gradient get multiply by learning rate each time, so it will keep reducing and become neglegible that means vanishing gradient problem.)

model.add(Dense(128, activation = "relu"))  # 128   :  784x128+ 128*(1bias)   >Total Parameters = 784x128 + 128

# We will have one output Layer
# We have 10 classes(categories) in output feature. So we will have 10 neurons in Output Layer.
# We have to use softmax as activation function beacuse we have multiclass classification project(problem).

model.add(Dense(10, activation = "softmax"))  # 128 is attached with 10 neurons= 128x10=1280 +10*(1bias)  >Total Parameters = 128*10+10
````

**Types of Loss Function** :


**1.Log Loss (Binary Cross-Entropy)** :

- Log Loss (also called Binary Cross-Entropy Loss) is a loss function used for binary classification problems, where there are only two possible classes.

- Log Loss measures how different the predicted probability is from the actual answer. Lower log loss means better predictions.

- It compares the predicted probability with the actual class.


#### Why is it called "Log Loss"?

- It uses a logarithm (log) in its calculation.

- The logarithm gives:
- A small penalty for correct, confident predictions.
- A very large penalty for confident but wrong predictions.

- This encourages the model to make accurate and well-calibrated probability predictions.

- The binary log loss formula is:
 
       (ylog(p)+(1−y)log(1−p))
       
       
       
       
**2.Cross-Entropy Loss** : 

- A generalized form of the log loss, which is used for multi-class classification problems.

- Cross-Entropy Loss is a loss function used for classification problems. It measures how different the model's predicted probabilities are from the correct class.

- Cross-Entropy Loss is a loss function that measures the difference between the predicted probabilities and the actual class labels. The smaller the loss, the better the prediction.



**Optimizer** :

- The optimizer is an algorithm that helps a neural network learn by updating its weights and biases so that the loss (error) becomes smaller.

- Optimizers are algorithms or methods used to change the attributes of your neural network such as weights and learning rate in order to reduce the losses.

- The loss function calculates how big the mistake is.

- Now the model needs to improve. The optimizer decides how to change the weights so the next prediction is better.

- Without an optimizer, the neural network would know it made a mistake but would not know how to fix it.

**Total params:** 
   - The sum of all parameters, both trainable and non-trainable.

**Trainable params:** 
   - Trainable parameters are the weights and biases that the CNN model is allowed to learn and update(change) during training.
   - (Parameters that the model will adjust during training.)

**Non-trainable params:** 
   - Non-trainable parameters are the weights and biases that the CNN model is not allowed to learn and update(change) during training.
   - Parameters that do not change during training.

**Optimizer params:** 
   - Parameters used by the optimizer, which may include internal state or configurations.
   - optimizer parameters are the settings that control how the model updates its trainable parameters (weights and biases) during training.
        - Diiferent types of optimizer parameters : 
            - 1. **Learning rate** :
                - This is the most important one.
                - It controls how big each update should be.
                
            - 2. **Momentum** :
                - Used in optimizers such as SGD.
                - It helps the model continue moving in a useful direction instead of changing direction too quickly.
                
            - 3. **Weight decay** :
                - Helps prevent the model from becoming too complex and can reduce overfitting.

## Evaluate on Unseen Data

**1. Loading the Image:**

we use PIL (Python Imaging Library) to open the image. This image can be in various formats like JPG, PNG, etc.

**2. Resizing the Image to 28x28:**

Since we trained your model on 28x28 images, you're resizing the input image to this specific size. If the input images during training were also grayscale, the dimensions would be (28, 28).

**3. Converting Image to Numpy Array:**

This converts the resized image into a NumPy array. The result is a 3D array with dimensions (28, 28, 3) because your original image is likely in RGB format (3 channels).

**4. Converting to Grayscale:**

Here, ywe are averaging the pixel values across the 3 color channels (Red, Green, and Blue). This operation reduces the array from a shape of (28, 28, 3) to (28, 28), effectively converting it into a grayscale image.

**5. Scaling the Image:**

we are normalizing the pixel values by dividing by 255. In this case, pixel values are in the range of 0 to 255, and dividing by 255 scales them to a range between 0 and 1, which is common in neural networks for better convergence.

**6. Visualizing the Scaled Image:**

This uses matplotlib to display the scaled grayscale image. It's useful for verifying that the image processing is correct.

**7. Reshaping the Image:**

Here, you are reshaping the image to add a batch dimension. The expected input to most neural networks is a 4D tensor where the shape is (batch_size, height, width, channels).

Since you trained your model with 28x28 grayscale images, you reshape it to (1, 28, 28) where:

- 1 is the batch size (since you're testing with one image).

- 28, 28 is the height and width of the image.

No channel dimension is specified because it's a single-channel grayscale image.
